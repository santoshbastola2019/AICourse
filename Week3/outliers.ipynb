```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/bad_daraz_dataset.csv")
df
```

```python
df["price"] = pd.to_numeric(df["price"], errors = "coerce")
df
```

```python
# Calculate Q1 and Q3
q1 = df["price"].quantile(0.25)
q3 = df["price"].quantile(0.75)
median = df["price"].quantile(0.50)

iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
lower_bound
lower_bound
median
```

```python
outliers = df[(df["price"] < lower_bound) | (df["price"] > upper_bound)]
outliers
```

```python
len(outliers)
```

```python
# Handling outlier using the median value
df.loc[(df["price"] < lower_bound) | (df["price"] > upper_bound), "price"] = median
df
```

```python
# Handling oulier
# Using Winsorization
df["price"].clip(lower_bound, upper_bound) 
```

```python
# USing Z-score
df = pd.read_csv("../datasets/bad_daraz_dataset.csv")
df
```

```python
df["price"] = pd.to_numeric(df["price"], errors = "coerce")
df
```

```python
mean = df["price"].mean()
std = df["price"].std()

# mean 13007.692307692309
#std 27605.71986438262

df["z_score"] = (df["price"] - mean) / std # z-score
outliers = df[df["z_score"].abs() > 2]
outliers
```
