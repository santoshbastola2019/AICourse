```python
# Vectorization
numbers = np.array([1, 2, 3, 4, 5])
# 10, 10, 10, 10, 10 [1, 2, 3, 4, 5] --> 1 * 10, 2 * 10, 3 * 10, 4 * 10, 5 * 10 (Behind the scene) => this is done in broadcasting

result = numbers * 10 
# result
result[(result > 30) & (result < 50)] # Filtering in vectorization
# result[result > 30] # array([False, False, False,  True,  True])
# print(result)
            
                
```
