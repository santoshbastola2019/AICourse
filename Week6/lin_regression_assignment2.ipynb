```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("../datasets/student_multiple_features.csv")
df.head(5) # display first five rows
df.tail(5) # display five last rows 
df.shape # display the shape of the dataset
df.info() # display information of dataset
df.describe() # display statistical summary
df
```

```python
# Handling Missing values
df.isnull().sum()
df["PreviousMarks"] = df["PreviousMarks"].fillna(df["PreviousMarks"].mean())
df["Attendance"] = pd.to_numeric(df["Attendance"], errors = "coerce")
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df                              
```

```python
features = ["StudyHours", "Attendance", "PreviousMarks"]
target = "FinalMarks"

X = df[features].to_numpy()
y = df[target].to_numpy()

# Split to 80% and 20%
split = int( len(df) * 0.8)

# Training data
X_train = X[:split]
y_train = y[:split]

# Create Model
model = LinearRegression()
model.fit(X_train, y_train)

# Calculate
print("Bias:", model.intercept_)
print("Coefficient of StudyHours", model.coef_[0])
print("Coefficient of Attendance", model.coef_[1])
print("Coefficient of PreviousMarks", model.coef_[2])

# Prediction
features = [[5, 10, 70],
           [7, 95, 82],
           [3, 75, 55]
           ]

predictions = model.predict(features)

print("Final Marks of Student A", predictions[0])
print("Final Marks of Student B", predictions[1])
print("Final Marks of Student C", predictions[2])

```
