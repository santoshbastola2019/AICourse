```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("../datasets/student_marks.csv")
df.head(5) # display first five rows of dataset
df.shape # display shape of the dataset 
df.dtypes # display datatypes
```

```python
# Finding missing values
df.isnull().sum() # no missing values found
```

```python
features = "StudyHours"
target = "FinalMarks"

X = df[[features]].to_numpy()
y = df[target].to_numpy()

# Split to 80% and 20%
split = int(len(df) * 0.8)

# Training data
X_train = X[:split]
y_train = y[:split]

# Testing data
X_test = X[split:]
y_test = y[split:]

model = LinearRegression()
model.fit(X_train, y_train)

# Calculate bias and weight
print("Intercept:" , model.intercept_)
print("Weight:", model.coef_[0])

# Marks prediction
hours = np.array([[2], [4], [6]])

predictions = model.predict(hours)

print("Predicted marks for 2 hours:", predictions[0])
print("Predicted marks for 4 hours:", predictions[1])
print("Predicted marks for 6 hours:", predictions[2])
```
