```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
index = [1, 2, 3, 4] # x-axis
price = [1000, 2000, 3000, 4300] # y-axis
plt.figure(figsize=(6, 6))
plt.plot(index, price)
# plt.show() # It was required to show chart previously but nit now
```

```python
df = pd.read_csv("../datasets/daraz_dataset.csv")
category = df.groupby("Category")["Product"].count()
plt.figure(figsize=(10,6))
plt.bar(category.index, category.values, color = "green", edgecolor = "black")
plt.title("Products and Categories Bar graph")
plt.xlabel("Categories", fontsize = 20)
plt.ylabel("Number of Products", fontsize = 20)
plt.grid(axis = "y", linestyle = "--", alpha = 0.5)
```

```python
# Pie Chart
category = df.groupby("Category")["Product"].count()
plt.figure(figsize=(12, 6))
colors = ["#0000", "brown", "blue", "red", "orange", "purple", "pink"]
plt.pie(category.values, labels=category.index, colors = colors, shadow = True, startangle = 90, explode = (0.1,0,0,0,0,0,0))
plt.title("Product distribution by Category")
plt.legend()
```

```python
gender_products = df.groupby("Gender")["Product"].count()
plt.figure(figsize=(12, 6))
plt.pie(gender_products.values, labels= gender_products.index, colors = ["pink", "green"], explode = (0.1, 0))
plt.title("Products by gender")
plt.legend()
```

```python
# Histogram
plt.hist(df["UnitPrice"], color = "green", edgecolor = "black", alpha = 0.7)
plt.title("Distribution of UnitPrice")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
# Histogram Custom Beans
bins = [0, 5000, 10000, 20000, 50000, 80000]
plt.hist(df["UnitPrice"],bins = bins, color = "green", edgecolor = "black", alpha = 0.7)
plt.title("Distribution of UnitPrice")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
# Scatter Plot
data = {
    "Age" : [20, 22, 24, 26, 28, 30],
    "TotalAmount" : [1000, 1200, 1500, 2000, 3000, 32000]
}
dfd = pd.DataFrame(data)
plt.scatter(dfd["Age"], dfd["TotalAmount"], color = "green")
plt.xlabel("Age")
plt.ylabel("TotalAmount")
plt.grid(True)
plt.title("Age vs Total AMount")
```

```python
# Scatter Plot in our data
plt.scatter(df["UnitPrice"], df["Rating"], s = 80, alpha = 0.5)
plt.xlabel('Unit Price')
plt.ylabel('Rating')
plt.grid(True)
```

```python
# Daraz dataset age and totalamount
plt.scatter(df["Age"], df["TotalAmount"], color = "green", s = 80)
plt.xlabel("Age")
plt.ylabel("TotalAmount")
plt.grid(True)
plt.title("Age vs Total Amount")
```
