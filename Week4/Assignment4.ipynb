```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Data import and info
df = pd.read_csv("../datasets/dp_lms_student_analytics_dataset.csv")
df.info()
```

```python
# 1. Line Charts
# Monthly student Enrollment

# Convert to datetime
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"])
df["Month"] = df["EnrollmentDate"].dt.month_name()
months = ["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
monthly_enrollment = df.groupby("Month")["StudentID"].count().reindex(months)

plt.figure(figsize=(12, 6))
plt.plot(monthly_enrollment.index, monthly_enrollment.values)
plt.title("Monthly Student Enrollment", pad = 20, fontsize = 20)
plt.xlabel("Month", labelpad = 15, fontsize = 15)
plt.ylabel("Number of Students", labelpad = 15, fontsize = 15)
plt.grid(True, linestyle = "--")
```

```python
# Monthly Revenue Trend
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"])
df["Month"] = df["EnrollmentDate"].dt.month_name()
months = ["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
monthly_revenue = df.groupby("Month")["Revenue"].sum().reindex(months)

plt.figure(figsize=(12, 6))
plt.plot(monthly_revenue.index, monthly_revenue.values, color = "green")
plt.title("Monthly Revenue Trend", pad = 20, fontsize = 20)
plt.xlabel("Month", labelpad = 15, fontsize = 15)
plt.ylabel("Total Revenue", labelpad = 15, fontsize = 15)
plt.grid(True, linestyle = "--")
```

```python
# Average course Progress by Month
df["Month"] = df["EnrollmentDate"].dt.month_name()
months = ["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
monthly_course_progress = df.groupby("Month")["CourseProgress"].mean().reindex(months)

plt.figure(figsize=(12, 6))
plt.plot(monthly_course_progress.index, monthly_course_progress.values, color = "red")
plt.title("Average Monthly Course Progress Trend", pad = 20, fontsize = 20)
plt.xlabel("Month", labelpad = 15, fontsize = 15)
plt.ylabel("Average Course Progress", labelpad = 15, fontsize = 15)
plt.grid(True, linestyle = "--")
```

```python
# Average Attendance percentage by Month
df["Month"] = df["EnrollmentDate"].dt.month_name()
months = ["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
monthly_average_attendance = df.groupby("Month")["AttendancePercentage"].mean().reindex(months)

plt.figure(figsize=(12, 6))
plt.plot(monthly_average_attendance.index, monthly_average_attendance.values, color = "orange")
plt.title("Average Monthly Attendance Percentage Trend", pad = 20, fontsize = 20)
plt.xlabel("Month", labelpad = 15, fontsize = 15)
plt.ylabel("Average Attendance Percentage", labelpad = 15, fontsize = 15)
plt.grid(True, linestyle = "--")
```

```python
# 2. Bar Charts
# Number of students in each course
student_numbers = df.groupby("Course")["StudentID"].count()
plt.figure(figsize=(12,6))
plt.bar(student_numbers.index, student_numbers.values, color = "green", edgecolor = "black")
plt.xlabel("Courses", fontsize = 15, labelpad = 15)
plt.ylabel("Number of Students", fontsize = 15, labelpad=15)
plt.title("Number of Students In Each Course", fontsize = 20, pad = 20)
plt.grid(axis="y", linestyle = "--", alpha = 0.5)
```

```python
# Revenue generated by each course
course_revenue = df.groupby("Course")["Revenue"].sum()
plt.figure(figsize=(12,6))
plt.bar(course_revenue.index, course_revenue.values, color = "skyblue", edgecolor = "black")
plt.xlabel("Courses", fontsize = 15, labelpad = 15)
plt.ylabel("Total Revenue", fontsize = 15, labelpad=15)
plt.title("Revenue Generated By Each Course", fontsize = 20, pad = 20)
plt.grid(axis="y", linestyle = "--", alpha = 0.5)
```

```python
# Average final score by course
course_avg_score = df.groupby("Course")["FinalScore"].mean()
plt.figure(figsize=(12, 6))
plt.bar(course_avg_score.index, course_avg_score.values,edgecolor = "black" ,color = "purple")
plt.title("Average Final Score By Course", pad = 20, fontsize = 20)
plt.xlabel("Courses", labelpad = 15, fontsize = 15)
plt.ylabel("Average final score", labelpad = 15, fontsize = 15)
plt.ylim(73, 75)
plt.grid(axis="y", linestyle="--", alpha=0.5)
```

```python
# Final Attendance percentage by course
course_avg_attendance = df.groupby("Course")["AttendancePercentage"].mean()
plt.figure(figsize=(12, 6))
plt.bar(course_avg_attendance.index, course_avg_attendance.values,edgecolor = "black" ,color = "orange")
plt.title("Average Attendance Percentage By Course", pad = 20, fontsize = 20)
plt.xlabel("Courses", labelpad = 15, fontsize = 15)
plt.ylabel("Average Attendance Percentage", labelpad = 15, fontsize = 15)
plt.ylim(70, 80)
plt.grid(axis="y", linestyle="--", alpha=1, color = "violet")
```

```python
# Number of Students from each city
students_by_city = df.groupby("City")["StudentID"].count()
plt.figure(figsize=(12, 6))
plt.bar(students_by_city.index,students_by_city.values, color = "yellow", edgecolor = "black")
plt.title("Number of Students From Each City", pad = 20, fontsize = 20)
plt.xlabel("City", labelpad = 15, fontsize = 15)
plt.ylabel("No of Students", labelpad = 15, fontsize = 15)
plt.grid(axis="y", linestyle = "--", alpha = 0.6)
```

```python
# Number of students under each instructor
students_by_instructors = df.groupby("Instructor")["StudentID"].count()
plt.figure(figsize=(10, 6))
plt.bar(students_by_instructors.index,students_by_instructors.values, color = "grey", edgecolor = "black")
plt.title("Number of Students Under Each Instructor", pad = 20, fontsize = 20)
plt.xlabel("Instructor", labelpad = 15, fontsize = 15)
plt.ylabel("No of Students", labelpad = 15, fontsize = 15)
plt.grid(axis="y", linestyle = "--", alpha = 0.6)
```

```python
# 2. Horizontal Bar charts
top10_course_by_enrollment = df.groupby("Course")["StudentID"].count().sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 7))
plt.barh(top10_course_by_enrollment.index, top10_course_by_enrollment.values, color = "purple", edgecolor = "black")
plt.xlabel("Number of Enrollments", fontsize=15, labelpad=15)
plt.ylabel("Courses", fontsize=15, labelpad=15)
plt.title("Top 10 Courses by Enrollment", fontsize=20, pad=20)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.gca().invert_yaxis() 
```

```python
# Top 10 instructors by student enrollment
top10_instructors_by_enrollment = df.groupby("Instructor")["StudentID"].count().sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 7))
plt.barh(top10_instructors_by_enrollment.index, top10_instructors_by_enrollment.values, color = "lavender", edgecolor = "black")
plt.xlabel("Number of Enrollments", fontsize=15, labelpad=15)
plt.ylabel("Instructors", fontsize=15, labelpad=15)
plt.title("Top 10 Instructors by Enrollment", fontsize=20, pad=20)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.gca().invert_yaxis() 
```

```python
# Top 10 students with the higher final scores
stundents_highest_score = df[["StudentName","FinalScore"]].sort_values(by = "FinalScore", ascending = False)
top10_stundents_highest_score = stundents_highest_score.head(10)
top10_stundents_highest_score
plt.figure(figsize=(12, 7))
plt.barh(top10_stundents_highest_score["StudentName"], top10_stundents_highest_score["FinalScore"], color = "green", edgecolor = "black")
plt.xlabel("FinalScore", fontsize=15, labelpad=15)
plt.ylabel("StudentName", fontsize=15, labelpad=15)
plt.title("Top 10 Students With Highest Final Score", fontsize=20, pad=20)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.xlim(90, 100)
plt.gca().invert_yaxis() 
```
