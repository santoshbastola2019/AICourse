```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("../datasets/decisiontree_classification.csv")

df["Result"] = df["Result"].map({
    "Fail" : 0, 
    "Good" : 1, 
    "Average" : 2, 
    "Excellent" : 3
})

X = df[["StudyHours"]]
y = df["Result"]

model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

#model train 
model.fit(X,y)
print(model.tree_.threshold)
new_students = pd.DataFrame({
    "StudyHours" : [2,3,5]
})
predictions = model.predict(new_students)
print(predictions)
```

```python
df = pd.read_csv("../datasets/student_multiclass_multiple_features.csv")
df["Result"] = df["Result"].map({
    "Fail" : 0, 
    "Average" : 1, 
    "Good" : 2
})

X = df[["StudyHours","Attendance","PreviousMarks"]]
y = df["Result"]

model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)
#model train 
model.fit(X,y)
# print(model.tree_.threshold)
new_students = pd.DataFrame({
    "StudyHours" : [5], 
    "Attendance" : [70],
    "PreviousMarks" : [65]
    
})
predictions = model.predict(new_students)
print(predictions)
```
